# How to create a dataframe from pandas and numpy

```python
#creating from Arrays
test_scores= np.array([[2,3,5,0],[3,5,1,2],[5,6,4,5],[8,9,6,7],[3,5,2,5]])
test_scores
```

## create a dataframe with pandas and Numpy

```python
show_data= pd.DataFrame(test_scores,)
show_data
```

modify_df= pd.DataFrame(test_scores, index=[0] , columns=[])

```python
modify_df= pd.DataFrame(test_scores, index=['Emeka','victor','helen','chisom','prosper'] ,
                        columns=['maths','phy','chem','bio'])
```

```python
modify_df
```

## Creating a Dataframe from Dictionary

```python
import pandas as pd
```

```python
subject=['maths','phys', 'chem','bio','Eng']
```

```python
scores=[89,70,67,78,59]
```

```python
My_dict= { 'subject':subject, 'Scores':scores}
```

```python
my_df= pd.DataFrame(My_dict)
my_df
```

```python
subject=['maths','phys', 'chem','bio','Eng']
scores=[89,70,67,78,59]
subject.append('TOTAL')
scores.append(sum(scores))
```

```python
My_dict= { 'subject':subject, 'Scores':scores}
my_df=pd.DataFrame(My_dict)
```

```python
my_df
```
